```python
#Bubble sort-----First simple example

print("\n\nBUBBLE SORT")
print("=" *40)
numbers=[5,3,8,1,2]
print("before numbers:",numbers)
n=len(numbers)
```

```python
for i in range(n):
    
    #inner loop comparing neighbouring elements
    for j in range(0,n-i-1):
        
       #if left number is bigger
        if numbers[j]>numbers[j+1]:
            
           #swap
            numbers[j],numbers[j+1]=numbers[j+1],numbers[j]


print("After:",numbers)
```

```python
 # ============================================================
# PART 5: BUBBLE SORT - SHOW EVERY PASS
# ============================================================
print("\n\nBUBBLE SORT - STEP BY STEP")
print("=" * 40)
numbers = [5, 3, 8, 1, 2]
print("Starting list:", numbers)
n = len(numbers)
for i in range(n):
    print("\nPass", i + 1)
    for j in range(0, n - i - 1):
        print("Compare:", numbers[j], "and", numbers[j + 1])
        if numbers[j] > numbers[j + 1]:
            print("SWAP")
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
        else:
            print("NO SWAP")
        print("List:", numbers)
print("\nFinal sorted list:", numbers)
```

```python
# PART 10: INSERTION SORT
# ============================================================
print("\n\nINSERTION SORT")
print("=" * 40)
numbers = [5, 2, 4, 6, 1, 3]
print("Before:", numbers)
for i in range(1, len(numbers)):
    # Take current value
    key = numbers[i]
    # Compare with previous values
    j = i - 1
    while j >= 0 and numbers[j] > key:
        numbers[j + 1] = numbers[j]
        j = j - 1
    # Insert key into correct position
    numbers[j + 1] = key
    print("After inserting", key, ":", numbers)
print("Final:", numbers)
```
