```python
#Linesr search
def linear_search(numbers,target):

    comparisons=0

    #check every element
    for i in range(len(numbers)):

        comparisons += 1

        if numbers[i]==target:
            print("found!")
            print("comparisons:",comparisons)
            return i
            print("Not found")
            print("comparisons:",comparisons)

    #target not found
    return -1
```

```python
numbers=[10,20,35,30,45]
print(linear_search(numbers,45))
```

**Binary search**


```python
numbers=[10,20,30,40,50,60,70]
```

```python
target= 50
first=0
last=len(numbers)-1

while first <= last:

    mid=(first+last)//2

    if numbers[mid]== target:
        print("found at index: ",mid)
        break

    elif numbers[mid]<target:
        first=mid+1
    else:
        last=mid-1
```
