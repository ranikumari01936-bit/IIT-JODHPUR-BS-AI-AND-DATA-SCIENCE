```python
# PRACTICAL 2: TWO-PHASE LOCKING (2PL)
# ============================================================
print("PRACTICAL 2: TWO-PHASE LOCKING")
# Create a transaction schedule
schedule = [
    "Lock(A)",
    "Lock(B)",
    "Lock(C)",
    "Unlock(A)",
    "Unlock(B)",
    "Unlock(C)"
]
print("\nTransaction Schedule:")
for step in schedule:
    print(step)
# Check whether the schedule follows 2PL
unlock_started = False
valid = True
for step in schedule:
    # Check if an unlock operation has started
    if "Unlock" in step:
        unlock_started = True
    # No new lock is allowed after unlocking starts
    if "Lock" in step and unlock_started:
        valid = False
print("\nIs this a valid 2PL schedule?")
if valid:
    print("YES - Valid Two-Phase Locking")
else:
    print("NO - 2PL Rule Violated")
```

```python
# ============================================================
# PRACTICAL 3: INVALID 2PL SCHEDULE
# ============================================================

print("\n===================================")
print("INVALID 2PL EXAMPLE")
print("===================================")

schedule = [
    "Lock(A)",
    "Lock(B)",
    "Unlock(A)",
    "Lock(C)"
]

print("\nTransaction Schedule:")
```

```python
for step in schedule:
    print(step)


# Validate the schedule

unlock_started = False
valid = True

for step in schedule:

    if "Unlock" in step:
        unlock_started = True

    if "Lock" in step and unlock_started:
        valid = False

if valid:
    print("Valid 2PL")
else:
    print("Invalid 2PL - New lock acquired after unlocking")
```

```python
# ============================================================
# PRACTICAL 4: DEADLOCK DETECTION USING NETWORKX
# ============================================================
print("\n===================================")
print("PRACTICAL 4: DEADLOCK DETECTION")
print("===================================")
# Install NetworkX if required in Google Colab
!pip -q install networkx
# Import required libraries
import networkx as nx
import matplotlib.pyplot as plt
# Create a directed graph
G = nx.DiGraph()
# Create waiting relationships between transactions
# T1 waits for T2
# T2 waits for T3
# T3 waits for T1
edges = [
    ("T1", "T2"),
    ("T2", "T3"),
    ("T3", "T1")
]
G.add_edges_from(edges)
# Draw the wait-for graph
plt.figure(figsize=(5, 4))
nx.draw(
    G,
    with_labels=True,
    node_size=2000,
    font_size=12,
    arrows=True
)
plt.title("Deadlock Wait-For Graph")
plt.show()
```

```python
# Detect circular waiting

try:

    cycle = nx.find_cycle(G)

    print("DEADLOCK DETECTED!")

    print("Circular waiting:", cycle)

except nx.NetworkXNoCycle:

    print("No Deadlock Detected")
```

```python
# ============================================================
# PRACTICAL 5: NO DEADLOCK EXAMPLE
# ============================================================

print("\n===================================")
print("NO DEADLOCK EXAMPLE")
print("===================================")

G2 = nx.DiGraph()

# No circular dependency

G2.add_edges_from([
    ("T1", "T2"),
    ("T2", "T3")
])


# Visualize the graph

plt.figure(figsize=(5, 4))

nx.draw(
    G2,
    with_labels=True,
    node_size=2000,
    arrows=True
)

plt.title("No Deadlock Graph")

plt.show()


# Check for deadlock

try:

    cycle = nx.find_cycle(G2)

    print("Deadlock Detected")

except nx.NetworkXNoCycle:

    print("No Deadlock - No Circular Waiting")
```
