```python
# ============================================
# SEGMENT 1: Environment Setup
# ============================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize']=(8,5)
```

```python
print("pandas version:",pd.__version__)
print("Numpy version:",np.__version__)
print("seaborn version:",sns.__version__)

```

```python
# ============================================
# Load Iris directly from scikit-learn
# ============================================
from sklearn.datasets import load_iris
iris=load_iris()
iris
```

```python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df.head()
```

```python
# iris is a "Bunch" object — dict-like, with data, target, feature_names, target_names
print("Keys available:", iris.keys())
print("\nFeature names:", iris.feature_names)
print("Target names (species):", iris.target_names)
```

```python
print("\nData shape:", iris.data.shape)
print("Target shape:", iris.target.shape)
```

```python
df['species']=iris.target
df['species']=df['species'].map({0:'setosa',1:'versicolor',2:'virginica'})
df.head()
```

```python
# Clean up column names (remove " (cm)" suffix for convenience)
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
df.head(6)
df.tail(6)
```

  **Segment 3 — Descriptive Statistics**

```python
numeric_cols=['sepal_length','sepal_width','petal_length','petal_width']
df.describe()
```

```python
print("MEAN:\n", df[numeric_cols].mean())
print("\nMEDIAN:\n", df[numeric_cols].median())
print("\nVARIANCE:\n", df[numeric_cols].var())
print("\nSTANDARD DEVIATION:\n",df[numeric_cols].std())
```

```python
print("Skewness (asymmetry of distribution):\n", df[numeric_cols].skew())
```

```python
print("\nKurtosis (tailedness of distribution):\n", df[numeric_cols].kurt())
```

```python
# Percentiles / Quantiles
percentiles = df[numeric_cols].quantile([0.1, 0.25, 0.5, 0.75, 0.9])
print("Custom percentiles:")
percentiles
```

```python
# Grouped descriptive statistics by species
grouped = df.groupby('species')[numeric_cols].agg(['mean', 'median', 'var', 'std'])
grouped
```

```python
# Coefficient of Variation (std/mean) — normalizes spread for comparison across features
cv = df[numeric_cols].std() / df[numeric_cols].mean()
print("Coefficient of Variation (lower = more consistent feature):")
cv.sort_values()
```

```python
# Correlation matrix
corr_matrix = df[numeric_cols].corr()
print("Correlation matrix:")
corr_matrix
```

```python
sns.heatmap(df[numeric_cols].corr(), annot=True)
```

```python
# Covariance matrix
cov_matrix = df[numeric_cols].cov()
print("Covariance matrix:")
cov_matrix
```

```python
# --- Hands-on Exercise 2 ---
# "Which feature has the highest variance?
#Which species has the most  consistent (lowest variance) petal_length?"
print("Highest variance feature overall:")
print(df[numeric_cols].var().idxmax())
```

```python
print("\nLowest petal_length variance by species:")
print(df.groupby('species')['petal_length'].var().idxmin())
```
