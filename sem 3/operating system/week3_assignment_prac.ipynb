```python
import threading
import time

def deliver_order(order_number):

    #get the name of the cureently running thread

    agent_name=threading.current_thread().name
    print(f"{agent_name} is picking up Order {order_number}")

    
     # Simulate delivery time
    time.sleep(50)
    print(f"{agent_name} has delivered Order {order_number}")
    
```

```python
# Create 5 threads
threads = []

for i in range(1, 6):
    thread = threading.Thread(
        target=deliver_order,
        args=(i,),
        name=f"Agent-{i}"
    )
    threads.append(thread)

    
```

```python
# Start all threads
for thread in threads:
    thread.start()


# Wait for all threads to finish
for thread in threads:
    thread.join()


# Final message
print("All 5 orders delivered!")
```
