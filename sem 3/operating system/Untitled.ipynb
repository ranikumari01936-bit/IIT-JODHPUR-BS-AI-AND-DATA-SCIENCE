```python
import psutil

print("PID\tProcess Name")

count = 0

for process in psutil.process_iter(['pid', 'name']):
    try:
        print(process.info['pid'], "\t", process.info['name'])
        count += 1

        if count == 10:
            break

    except (psutil.NoSuchProcess, psutil.AccessDenied):
        pass
```

```python
import psutil

print("PID\tProcess Name\tStatus")

count = 0

for process in psutil.process_iter(['pid', 'name', 'status']):
    try:
        info = process.info

        print(
            info['pid'],
            "\t",
            info['name'],
            "\t",
            info['status']
        )

        count += 1

        if count == 10:
            break

    except (psutil.NoSuchProcess, psutil.AccessDenied):
        pass
```
